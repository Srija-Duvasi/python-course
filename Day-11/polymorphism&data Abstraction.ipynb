# polymorphism


```python
# the word polymorphism means "Many forms"
# in polymorphism, same function or same operator can perform many different task.

# Type of polymorphism:
# 1. Function overloading
# 2. Operator overloading
# 3. Method overriding
```

# 1. Function overloading


# Using default arguments for function overloading

```python
# Ex-1:
def fun(var1,var2 = None,var3 = None):
    if var2 is not None:
        print(var1,var2,var3)
    else:
        print(var1)
fun(10)
fun(10,20,30)
```

```python
# Ex-2:default argument
def calculator(a = 0,b = 0,c = 0,d = 0):
    return a + b + c + d

print(calculator())
print(calculator(100))
print(calculator(100,200))
print(calculator(100,200,300))
print(calculator(100,200,300,400))
```

```python
# *args for function overloading

def calculator(*args):
    return sum(args)

print(calculator())
print(calculator(100))
print(calculator(100,200))
print(calculator(100,200,300))
print(calculator(100,200,300,400))
print(calculator(100,200,300,400,500,600,700,800))
```

```python
# function overloading using decorator
from multipledispatch import dispatch

@dispatch(int)
def fun(var1):
    print(var1)

@dispatch(int, int)
def fun(var1, var2):
    print(var1, var2)

@dispatch(float, float,int)
def fun(a, b, c):
    print(a, b, c)

fun(10)
fun(20, 30)
fun(1.2, 3.4, 5)
```

# 2. Operator overloading


```python
# we will perform different operations on the user defined objects using operators.
class dosa:
    def __init__(self, variety,price):
        self.variety = variety
        self.price = price

    def __add__(self, d2_val):
        total_price = self.price + d2_val.price
        return total_price

    def __sub__(self, d2_val):
        total_price = self.price - d2_val.price
        return total_price

    def __mul__(self, d2_val):
        total_price = self.price * d2_val.price
        return total_price

    def __truediv__(self, d2_val):
        total_price = self.price / d2_val.price
        return total_price

    def __floordiv__(self, d2_val):
        total_price = self.price // d2_val.price
        return total_price

    def __mod__(self, d2_val):
        total_price = self.price % d2_val.price
        return total_price

    def __pow__(self, d2_val):
        total_price = self.price ** d2_val.price
        return total_price

d1 = dosa("Masala", 60)
d2 = dosa("Egg", 80)

print(d1 + d2) # __add__
print(d1 - d2) # __sub__
print(d1 * d2) # __mul__
print(d1 / d2) #__truediv__
print(d1 // d2) #__floordiv__
print(d1 % d2) #__mod__
# print(d1 ** d2) #__pow__
```

```python
# Comparision operator overloading

class dosa:
    def __init__(self, variety,price):
        self.variety = variety
        self.price = price

    def __gt__(self, d2_val):
        total_price = self.price > d2_val.price
        return total_price

    def __lt__(self, d2_val):
        total_price = self.price < d2_val.price
        return total_price

    def __eq__(self, d2_val):
        total_price = self.price == d2_val.price
        return total_price

    def __ge__(self, d2_val):
        total_price = self.price >= d2_val.price
        return total_price
        
    def __le__(self, d2_val):
        total_price = self.price <= d2_val.price
        return total_price

d1 = dosa("Masala", 60)
d2 = dosa("Egg", 80)

print(d1 > d2) # __gt__
print(d1 < d2) # __lt__
print(d1 == d2) # __eq__
print(d1 >= d2) #__ge__
print(d1 <= d2) #__le__

```

# Data Abstraction


```python
# def: Abstraction hides internal implementation details and shows only essential features.

# why use data  Abstraction:
1.To hide the complex details
2.this make the code more readable and secure

# libraries needed for Data Abstraction
from abc import ABC, abstractmethod

# Rules of data abstraction in python:
1. Abstract class can have both abstract and concrete method
2. you cannot instantiate a object from abstract class
3. all subclasses must implement ALL abstract methods.

# abstract method: a method declared but not defined/implemented/without code.
                : abstarct method is defined using @abstractmethod decorator
# concrete method: This method have the body part
# concrete method need not be defined in child class
```

```python
#Example-1: how to define abstract class, abstract method

from abc import ABC, abstractmethod

#abstract class
class Head_Mahindra(ABC):
    @abstractmethod
    def air_bag(self): #abstract method
        pass

    @abstractmethod
    def LED_Screen(self):
        pass

    def insurance(self): #concrete
        print("Please take Mahindra company insurance")
        

class hyd_mahindra(Head_Mahindra): # chid class
    def air_bag(self): #abstract method
        print("I will use Nylon to design the Air bag")

    def LED_Screen(self):
        print("Sony LED")

class noida_mahindra(Head_Mahindra):
    def air_bag(self): #abstract method
        print("I will use Cotton to design the Air bag")

    def LED_Screen(self):
        print("Samsung LED")


obj_hyd = hyd_mahindra() # create object of hyd child class
obj_noida = noida_mahindra()

obj_hyd.insurance()
# obj_head = Head_Mahindra()
```
