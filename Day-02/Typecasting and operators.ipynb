# Type Casting

```python
#integer typecasting : int()
v1 = int(5.6) #float to int
v2 = int("300") #string to int
v3 = int(True) #bool to int
v4 = int(False) #bool to int
v5 = int("1010", 2) #binary to int

print(v1, type(v1))
print(v2, type(v2))
print(v3, type(v3))
print(v4, type(v4))
print(v5, type(v5))
```

```python
# covert to float : float()
v1 = float(5) #int to float
v2 = float("40") #String to float
v3 = float(True) #bool to float

print(v1, type(v1))
print(v2, type(v2))
print(v3, type(v3))
```

```python
# covert to string : str()
v1 = str(20)
v2 = str(1.9)
v3 = str(True)

print(v1, type(v1))
print(v2, type(v2))
print(v3, type(v3))
```

```python
# covert to boolean : booll()
v1 = bool(20)
v2 = bool(1.9)
v3 = bool(True)
v4 = bool("Hello")
v5 = bool(0)

print(v1, type(v1))
print(v2, type(v2))
print(v3, type(v3))
print(v4, type(v4))
print(v5, type(v5))
```

# Operators

#### 1. Arithmatic operators(+,-,*,/,%,//,**)

```python
a = 10
b = 3

print(a + b)
print(a - b)
print(a*b)
print(a/b) #decimal division
print(a//b) #floor value
print(a%b) #remainder
print(a**b) #power
```

```python
a = -10
b = 3

print(a//b) #// it takes smaller value as -4 is less than -3 ans will be -4
```

### 2. Bitwise operators(<<, >>, ~, &, |,^)¶

```python
#bitwise operators works on binary number
#bitwise operators are faster than arithmatic operators
#bitwise operators perform many operation in easy way due to binary number system
```

```python
#ADD Operator (&) If both the bit are true answer is true
#OR Operator (|) If one of the bit is true answer is true
#XOR Operator (^) If both bit are differnt then answer is True (if we have even number of 1's output is 0, if we have odd number of 1 = 1)
```

```python
a = 10
b = 3

print(a & b)
print(a | b)
print(a ^ b)
```

```python
left shift(<<), Right shift(>>)
```

```python
# Def (left shift): Insert new bits (0) to the LSB (number is doubing)
# Def (Right shift): Delete the bits from the LSB (number is half)
```

```python
a = 10
print(a<< 1)
print(a << 2)
print(a << 3) #it works as (a * (2  power k))
print(a << 6) #(10 * (2**6))
```

```python
a = 10
print(a >> 1) #it works as (a // (2 power K))
print(a >> 2) # (10 // (2**2)
print(a >> 3)
print(a >> 4)
```

```python
bitwise not (~)
```

```python
#def (not) : flip the bit 
a = 12 #(if we take postive number output will be negative + 1)
print(~a)
```

```python
a = -12 #(if we take negative number output will be positive - 1)
print(~a)
```

# 3. Assigment operators (+=, -=, *=, /=,...)¶

```python
a = 10

a+=5 # a = a + 5
print(a)
a-=5 # a = a -5
print(a)
a*=5
print(a)
a/=5
print(a)
a%=5
print(a)

# &=, |=, ^=, <<=, >>=, :=(walrus operator) ...

a = 10
a &= 3
print(a)

a = 10
a |= 3
print(a)

a = 10
a ^= 3
print(a)

a = 10
a <<= 3
print(a)

a = 10
a >>= 3
print(a)

#Walrus Operator (:=)
#Walrus Operator will first assign the value to the variable then pass it to outer function
#on the go u assign the value and then print the value

print(a := 3)
```

### 4. Comparison Operators(==, >=, <=, !=, <, >)¶

```python
a = 10
b = 5

print(a == b)
print(a >= b)
print(a <= b)
print(a != b)
print(a < b)
print(a > b)
```

### 5. Logical operators(and, or, not)

```python
#and, or, not
a = True
b = False

print( a and b)
print( a or b)
print( not b)
```

### 6. Identity Operator(is, is not)

```python
a = [1,2,3]
b = [1,2,3]
c = a

print(a == b)

print(a is b) #Its False because their memory address will be different
print(a is c) #Its True because c is assgined as a
print(a is not b)
print(a is not c)
```

### 7. Membership Operators (in, not in)

```python
fruits = ('Mamidipandu', 'Aratipandu', 'Batai', 'panasapandu')

print('Batai' in fruits)
print('Batai' not in fruits)
```
