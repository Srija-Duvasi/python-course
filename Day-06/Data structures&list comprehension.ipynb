## Delete item from list(remove(),pop(),clear(),del())

```python
food_list = ["papu","Biryani","fried rice","samber","payasum","curd rice"]

# remove() : delete the item with given name
food_list.remove("fried rice")


print(food_list)
```

```python
# pop() is used to delete the item from a particular ibdex
food_list.pop(4)
print(food_list)
```

```python
# pop() without index will delete the last element
food_list.pop()
print(food_list)
```

```python
food_list = ["papu","Biryani","fried rice","samber","payasum","curd rice"]

#delete multiple element together
del food_list[1:4]

print(food_list)
```

```python
# clear() is used to delete all the items from the list
food_list.clear()
print(food_list)
```

```python
# delete the list itself
del  food_list 
```

# list comprehension

```python
# creating the list in a single line using a loop

L1 = [ i for i in range(1,11)]
print(L1)
```

```python
# creating the list in a single line using a loop

L1 = [ i for i in range(1,11)]
print(L1)
```

```python
L3= [i for i in range(1,11) if i%2==0]
print(L3)
```

```python
for i in range(1,11):
    if i%2==0:
        print(i)
```

```python
L3= [i for i in range(1,11) if i%2==0 if i%3==0]
print(L3)
```

```python
for i in range(1,11):
    if i%2==0:
        if i%3==0:
            print(i,end = " ")
```

```python
L4 = [i**2 if i%2==0 else i+2 for i in range(1,11)]
print(L4)
```

```python
for i in range(1,11):
    if i%2==0:
        print(i**2)
    else:
        print(i+2)
```

```python
L4 = [i**2 if i%2==0 else i+2 for i in range(1,11) if i>5]
print(L4)
```

```python
L4 = [i**2 if i%2==0 else i+2 for i in range(1,11) if i>5 or i<9]
print(L4)
```

```python
L4 = [i**2 if i%2==0 else i+2 for i in range(1,11) if i>5 and i<9]
print(L4)
```
