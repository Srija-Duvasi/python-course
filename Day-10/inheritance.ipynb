# Inheritance

```python
# Def: inheritance allows us to define a class that inherits  the properties 
   #  or functions of the parent class.
# parent class = Base class
# child class = derived class
```

```python
# Types of inheritance:
1.single inheritance : one parent one child
2.Multiple inheritance: many parents one child
3.multilevel inheritance: koduku -> nanna -> thatha -> muthatha
4.hierarichal inheritance: one parent many child
5.hybrid inheritance: many parent many child.combine any two or more inheritance
```

# 1.single inheritance : one parent one child

```python
# Example-1:structure of single inheritance
# parent class
class hyd_food:
    taste = "very very very spicy"

    def Biryani(self):
        print("very very very tasty Biryani")


# child class
class PistaHouse(hyd_food):
    pass

obj1_child = PistaHouse() # it is the object of child class

# child class object can access the properties & function of parent class.
print(obj1_child.taste)
obj1_child.Biryani()
```

```python
# Example-2: super() function
# super(): use to access the parent properties and function from
       # inside of the child class

# parent class
class hyd_food:
    taste = "very very very spicy" # property/class variable

    def Biryani(self):
        print("very very tasty Biryani")


# child class
class PistaHouse(hyd_food):
    def display(self):
        super().Biryani()
        print(super().taste)

obj1_child = PistaHouse() # it is the object of child class
obj1_child.display()
```

```python
# Example - 3: constructor function

class hyd_food:
    taste = "very very very spicy" # property/class variable

    def Biryani(self):
        print("very very very tasty Biryani")


# child class
class PistaHouse(hyd_food):
    def __init__(self):
        print("I am pista house")

obj1_child = PistaHouse() # it is the object of child class
```

```python
class hyd_food:
    taste = "very very very spicy" # property/class variable

    def __init__(self):
        print("very very tasty Biryani")


# child class
class PistaHouse(hyd_food):
    def __init__(self):
        print("I am pistahouse class")

obj1_child = PistaHouse() # it is the object of child class
# if child class Constructor function is not defined
# then child class object will call automatically parent class function
```

```python
class hyd_food:
    taste = "very very very spicy"# property/class variable
    
    def __init__(self):
        print("Father car is maruti")
        
    def Biryani(self):
        print("very very tasty biryani")


# child class
class PistaHouse(hyd_food):
    
    def __init__(self):
        print("sreeja car is BMW")
        

obj1_child = PistaHouse() 
```

```python
# Example-1: sequence in which parent are assign that matter
class A:
    def display(self):
        print("I am A class")
        
class B:
    def display(self):
        print("I am B class")
        
class C(B,A):
    pass

obj_C = C()
obj_C.display()
```

```python
# Example-3: super() : follow the MRO approach
# MRO : Method Resolution Order
class A:
    def display(self):
        print("I am A class")
        
class B:
    def show(self):
        print("I am B class")
        
class C(A,B):
    def hello(self):
        super().display()
        super().show()
        

obj_C = C()
obj_C.hello()

# MRO : Method Resolution Order
print([i.__name__ for i in C.__mro__])
```

```python
class A:
    def display(self):
        print("I am A class")
        super().display()
        
class B:
    def display(self):
        print("I am B class")
        
class C(A,B):
    def hello(self):
        super().display()
        

obj_C = C()
obj_C.hello()
```

```python
# Example - 1:

class grantGF:
    def Gold(self):
        print("1000 KG")
    
class grandfather(grantGF):
    def land(self):
        print("500 Acr")
    
class father(grandfather):
    def car(self):
        print("BMW")
    
class son(father):
    def Bike(self):
        print("DUKE Bike")
obj_son = son()
obj_father = father()
obj_father.land()
obj_son.Bike()
obj_son.car()
obj_son.land()
obj_son.Gold()
```

```python
# Example - 3: super() function
class grantGF:
    def Gold(self):
        print("1000 KG")
    
class grandfather(grantGF):
    def land(self):
        print("500 Acr")
    
class father(grandfather):
    def car(self):
        print("BMW")
    
class son(father):
    def Bike(self):
        print("DUKE Bike")
        super().car()
        super().land()
        super().Gold()

obj_son = son()
obj_son.Bike()
```

# 4.Hierarichal inheritance: one parent many child


```python
# many child one parent
# structure of the Hierarichal inheritance

class superhero:
    pass
    
class sinchan(superhero):
    pass
    
class Doreman(superhero):
    pass
```

```python
# Example - 1:
class superhero:
    def intro(self):
        print("Superhero powers")
    
class sinchan(superhero):
    def comedy(self):
        print("Himavari")
    
class Doreman(superhero):
    def pocket(self):
        print("Nobita friend")

obj_doreman = Doreman()
obj_doreman.intro()
```

```python
# Example - 2: constructor function in Hierarichal inheritance

class superhero:
    def __init__(self):
        print(" I am Superhero construcor function")
    
class sinchan(superhero):
    def __init__(self):
        print(" I am sinchan construcor function")
    
class Doreman(superhero):
    def __init__(self):
        print(" I am Doreman construcor function")

obj_doreman = Doreman()
```

```python
# Example - 3: super() function
class superhero:
    def intro(self):
        print("Superhero powers")
    
class sinchan(superhero):
    def comedy(self):
        print("Himavari")
    
class Doreman(superhero):
    def pocket(self):
        print("Nobita friend")
        super().intro()

obj_doreman = Doreman()
obj_doreman.pocket()

# Method resolution order
print([i.__name__ for i in sinchan.__mro__])
```

# 5.Hybrid inheritance: many parent many child


```python
# it combine any two or more inheritance
# structure of the Hybrid inheritance

class A:
    pass
    
class B(A):
    pass
    
class C(B):
    pass
    
class D(B):
    pass
    
class E(C,D):
    pass 

obj_E = E()
```

```python
# Example - 3 : super() function

class A:    
    A_var = "A Class"
    def A_output(self):
        print("this is A output")
        
class B(A):
    B_var = "B Class"
    def B_output(self):
        print("this is B output")

class C(B):
    C_var = "C Class"
    def C_output(self):
        print("this is C output")
        
class D(B):
    D_var = "D Class"
    def D_output(self):
        print("this is D output")

class E(C,D):      
    E_var = "E Class"
    def E_output(self):
        print("this is E output")
        super().D_output()
        super().C_output()
        super().B_output()
        super().A_output()
        
obj_E =E()
obj_E.E_output()
```
