# popular inbuilt function for tuple

```python
salary = (6,70,100,600,700,90,10,80,120)

print(tuple(reversed(salary)))
print(len(salary))
print(min(salary))
print(max(salary))
print(sum(salary))

sorted_list = tuple(sorted(salary, reverse = True))
print(sorted_list)

# sort() is inplace function
# inplace means sorting is done in the same data structure

# sorted() function return the list
# sort() function don't return anything
```

```python
l1 = [34,40,50,60,1,3,4,70]

l1.sort()
print(l1)
```

```python
# Basic interview  questions
Q1: set is mutable or immutable?
ANS: sets are mutable (insert and delete),
but its items are immutable(update not allowed.

Q2:what is the time complexity for insert and delete operations in set?
ANS:BC,AC = O(1),AC = O(n)
set store the data at the backend using hash table
hash table take O(1) on average to store and delete the data.

Q3:Duplicate elements are allowed in set or not?
ANS:YES, duplicate elements in set not allowed.

Q4:Heterogeneous elements are allowed in set or not?
ANS:YES
Q5:set is ordered or unordered?
ANS:set is unordered. their is no indexing in set.
```

```python
s1 = {10,10,10,20,"hello"}
print(s1)#in set duplicate elemets is not allowed
```

# creating different types of sets


```python
empty_set = set()

set2 = set([10,20,30])
set3 = set((10,20,30))

# frozenset is immutable (cannot perform insert,delete,update)
fro_set = frozenset([10,20,30,40,50])

nested_set = {(10,20,30),fro_set,"Hello",30} # tuple + frozenset can be store in the set 
#nested_set2 = {[10,20,30],"Hello",30} # not allowed to store list in set
#nested_set3= {{10,20,30},"Hello",30} #set inside set cannot be store

#IMP: Mutable items cannot be store inside the set
# only immutable items can be store in the set

print(empty_set)
print(set2)
print(set3)
print(fro_set)
print(nested_set)
```

# insert in set (add(),update())


```python
set1 = {10,20.5,"Hello"}

#add() function is used to insert one item in the set
set1.add("World")

print(set1)
```

```python
set1.add((10,20,30,40))

print(set1)
```

```python
# update() is used for inserting multiple items together

set1.update((1000,2000))

print(set1)
```

# delete operation on set(remove(),discard(),pop(),clear(),del()

```python
dinner_set = {"Bagara rice","Biryani","Papu","chapati","Fish","mutton","prawns"}

# remove() function delete the one item whose name is provided
# remove() will raise an ERROR if item don't exist
dinner_set.remove("prawns") # if item is not will show the error

print(dinner_set)
```

```python
# discard() function delete the one item whose namr is provided
# discard() DONOT raise ERROR if item don't exist
dinner_set.discard("Biryani")  # if item is not their it will not show the error

print(dinner_set)
```

```python
# pop() will delete any random item
dinner_set.pop()

print(dinner_set)
```

```python
# clear() will delete all the items in the set
dinner_set.clear()
print(dinner_set)
```

```python
del dinner_set # it will detele the whole set
```
