# Recursive functions

```python
# Def : a function calling itself directly or indirectly.
# termination condition is MUST in recurstion.
# in recursive call, we must change the parameter value.
```

```python
# example-1
def foo(i):
    if i < 1: # termination condition or base condition
        return
    print(i)
    foo(i-1)

foo(6)
```

```python
# example-2
def foo(i):
    if i < 1: # termination condition or base condition
        return
    foo(i-1)
    print(i)

foo(6) # based on the stack it is stored
```

# Decoretors

```python
Def : A Decorator is a function that takes another function as input,
      adds extra functionality and returns a modified function,
      without changing the original function code.
```

```python
def decorator_fun(func):
    def wrapper_fun(): # wrapper function add the extra functionality
        func()
        print("add cherry") # add extar function
        print("add chocolate")
    return wrapper_fun

def cake():
    print("take the flour")
    print("add eggs and other ingredient")
    print("put the cake in oven")


decorated_function = decorator_fun(cake)
decorated_function()
```

```python
# example - 2 : use of @ to represent the decorators 
def decorator_fun(func): # func = > cake()
    def wrapper_fun():
        print("buy the ingredients") # wrapper function add the extra functionality
        func()
        print("add cherry") # add extar function
        print("add chocolate")
    return wrapper_fun


@decorator_fun
def cake():
    print("take the flour")
    print("add eggs and other ingredient")
    print("put the cake in oven")

cake()
```

```python
# Multiple decorator for same function
def decorator_fun1(func): # func = > cake()
    def wrapper_fun():
        print("buy the ingredients") # wrapper function add the extra functionality
        func()
    return wrapper_fun
    
def decorator_fun2(func): # func = > cake()
    def wrapper_fun():
        func()
        print("add cherry") # add extar function
        print("add chocolate")
    return wrapper_fun
    
@decorator_fun1
@decorator_fun2
def cake():
    print("take the flour")
    print("add eggs and other ingredient")
    print("put the cake in oven")

cake()
```

```python
# Example - 1: structure of class

# class is a keyword, food is the name
# class variable : the value of class variable remaims SAME for all the objects
# class variable value cannot be change by object
# class variable value can be only change by class name
class food:
    food_name = "Biryani" # class variable


obj1 = food() # creating the object (obj) of food() class
print(obj1.food_name) # accessing the class variable

# class variable value CAN be only change by class name
food.food_name = "mutton"

obj2 = food()
print(obj2.food_name) #class object (obj2) can access the class data

# access the class variable with the class name
print(food.food_name)

# class variable value cannot be change by object
obj3 = food()
obj3.food_name = "pappu" # Here obj3 , is creating its own local variable with same name

print(obj3.food_name) # printing local variable to obj3
```

# Encapsulation


```python
# Def : Encapsulation refers to the binding of data (attributes)
#  and function into one unit (class).

# Encapsulation restrict the direct access of the data. 

#we can access the variables/data/attribute through object outside the class.
#we can access the variable/data/attribute through functions defined inside the class.

class Bike:
    No_tyre = 2 # class variable / Attribute / data

    def __init__(self, name,price):
        self.bike_name = name # instance variable
        self.price = price
        

    def display(self):
        print(self.No_tyre,self.bike_name,self.price)

obj1 = Bike("Duke",2.5)

obj1.display()

print(obj1.No_tyre)
print(obj1.bike_name)
print(obj1.price)
```

### Access specifiers / types of attributes in encapsulation


```python
# 1.public : these variables can be accessed outside the class.
             # public variabes do not start with underscore.
# 2.protected : these variables can be accessed only inside/child class.
               # protected variable not INTENDED to access outside the class. 
               # proteced variable start with SINGLE underscore.
# 3.private : these variables can be accessed only inside the class.
               # not INTENDED to access outside/child class.
               # private variables starts with double underscore
```

```python
class marvels:
    def __init__(self,name,power,weapon):
        self.name = name # public instance variable
        self._power = power # protected instance variable
        self.__Ayudham = weapon # private instance variable

    def show(self):
        print(self.name,self._power,self.__Ayudham)
        
obj1 = marvels("thor",1000,"Hammer")
obj1.show()

print(obj1.name)
print(obj1._power)
# protected variable can be access ouside the class, but you should not access.
# in python public,protected,private concept is used just for sign/warning
print(obj1._marvels__Ayudham) # obj._classname__privatevariable
```
