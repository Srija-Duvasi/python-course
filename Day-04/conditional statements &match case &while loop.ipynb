```python
# Example 2     Structure of if-else statements
a = 10
if a % 2 == 0:
    print("Even")
else:
    print("ODD")
```

```python
# Example 3     Structure of elif statements(if one condition fails then we go to next condition)

Biryani = input("Famous Hyderabadi biryani")

if Biryani == "Paradise":
    print("Spicy & tasty")
elif Biryani == "Shahdab":
    print("double masala")
elif Biryani == "road side":
    print("budget friendly")
else:
    print("PG Biryani")
```

```python
# replace elif with else: if
Biryani = input("Famous Hyderabadi biryani")

if Biryani == "Paradise":
    print("Spicy & tasty")
else:
    if Biryani == "Shahdab":
        print("double masala")
    else:
        if Biryani == "road side":
            print("budget friendly")
        else:
            print("PG Biryani")
```

# match case


```python
# Match case is used to replace multiple if else condition
# Match case is more powerful than switch case because match case works with variables & data Structurs
```

```python
# Example 1
num = int(input())

match num:
    case 10:
        print("you entered 10")
    case 20:
        print("you entered 20")
    case 30:
        print ("you entered 30")
    case _:
        print("something else")
```

```python
# Example 2 : passind data structure to match case

data = [1,2,3,4]
match data:
    case [4,3,2,1]:
        print("Hello")
    case [1,2,3,4]:
        print("world")
```

```python
# example
Biryani = input("Famous Hyderabadi biryani")
match Biryani:
    case "Paradise":
        print("Spicy & tasty")
    case "Shahdab":
        print("double masala")
    case "road side":
        print("budget friendly")
    case _:
        print("something else")
        
```

```python
#Example 3: if condition with match case

val = int(input())

match val:
    case x if x > 0:
        print("it is positive number")
    case y if y < 0:
        print("it is negative number")
    case _:
        print("it is a zero")
```

```python
# in while loop u have to increment r decrement the value if not it ewill be infinite loop
# initialization, condition, increment/decrement
```

```python
# example - 1

i = 1
while i <= 10:
    print(i)
    i +=1
```

```python
# Example number 2 : Break Statment
# break statement helps to come out of the loop

i = 1
while i <= 10:
    print(i)
    if i == 5:
        break   # break will help u to comeout of loop when i ==5
    i += 1
```

```python
i = 1
while i <= 10:
    i += 1
    print(i)
    if i == 5:
        break   # break will help u to comeout of loop when i ==5
```

```python
# Example number 3 : Continue Statment (if continue satement is true it skips that values )
# Continue will take the control to the start of the loop
# if continue will execute then further statement next lline will not execute
# continue will take u to starting of the looop

i = 1
while i < 10:
    i += 1
    if i < 5:
        continue
    print(i)
```

```python
# example : else statement with while loop
# else statement executes only if loop is successfully completed
# we use else to check if the loop is successfully completed r not

i = 1
while i <=10:
    print(i)
    i += 1

else:     # to check if the loop is successfully completed r not
    print("loop is successfully completed")
```
