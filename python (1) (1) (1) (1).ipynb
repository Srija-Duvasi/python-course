#### 2. Bitwise operators(<<, >>, ~, &, |,^) 

```python
#bitwise operators works on binary number
#bitwise operators are faster than arithmatic operators
#bitwise operators perform many operation in easy way due to binary number system
```

```python
#ADD Operator (&) If both the bit are true answer is true
#OR Operator (|) If one of the bit is true answer is true
#XOR Operator (^) If both bit are differnt then answer is True (if we have even number of 1's output is 0, if we have odd number of 1 = 1)
```

```python
a = 10
b = 3

print(a & b)
print(a | b)
print(a ^ b)
```

left shift(<<), Right shift(>>)

```python
# Def (left shift): Insert new bits (0) to the LSB (number is doubing)
# Def (Right shift): Delete the bits from the LSB (number is half)
```

```python
a = 10
print(a<< 1)
print(a << 2)
print(a << 3) #it works as (a * (2  power k))
print(a << 6) #(10 * (2**6))
```

```python
a = 10
print(a >> 1) #it works as (a // (2 power K))
print(a >> 2) # (10 // (2**2)
print(a >> 3)
print(a >> 4)
```

# match case

```python
# Match case is used to replace multiple if else condition
# Match case is more powerful than switch case because match case works with variables & data Structurs
```

```python
# Example 1
num = int(input())

match num:
    case 10:
        print("you entered 10")
    case 20:
        print("you entered 20")
    case 30:
        print ("you entered 30")
    case _:
        print("something else")
```

```python
# Example 2 : passind data structure to match case

data = [1,2,3,4]
match data:
    case [4,3,2,1]:
        print("Hello")
    case [1,2,3,4]:
        print("world")
```

```python
# example
Biryani = input("Famous Hyderabadi biryani")
match Biryani:
    case "Paradise":
        print("Spicy & tasty")
    case "Shahdab":
        print("double masala")
    case "road side":
        print("budget friendly")
    case _:
        print("something else")
        
```

```python
#Example 3: if condition with match case

val = int(input())

match val:
    case x if x > 0:
        print("it is positive number")
    case y if y < 0:
        print("it is negative number")
    case _:
        print("it is a zero")
```

```python
# we have negative and positive indexing
# string[start:end:step]  start = 0, end = n-1, step = 1

Heroine = "Samantha"

print(Heroine[4])
print(Heroine[-5])

print(Heroine[0:5])
print(Heroine[0:5:3])
print(Heroine[::])
print(Heroine[::2])
print(Heroine[6:3:1])   # if step size is positive we move to the right side & we r 

print(Heroine[6:3:-1])  # for -ve indexing we move left side and end = end - (-1) = end + 1
print(Heroine[-1 : -6])  # by default step size is 1 as +ve step we move right size & it print empty
print(Heroine[-1:-6:-1])
print(Heroine[-7::])     # by default step size is 1 as +ve step we move right size 
print(Heroine[-7::])

print(Heroine[0:-1:])    # combined +ve And -ve

print(Heroine[-10::])   # when we start at out of range we start from the last in ve
print(Heroine[-1:-6:-2])
print(Heroine[10:-6:])
print(Heroine[0:2:1] + Heroine[5::1])
```

#### Encapsulation

```python
# Def : Encapsulation refers to the binding of data (attributes)
#  and function into one unit (class).

# Encapsulation restrict the direct access of the data. 

#we can access the variables/data/attribute through object outside the class.
#we can access the variable/data/attribute through functions defined inside the class.

class Bike:
    No_tyre = 2 # class variable / Attribute / data

    def __init__(self, name,price):
        self.bike_name = name # instance variable
        self.price = price
        

    def display(self):
        print(self.No_tyre,self.bike_name,self.price)

obj1 = Bike("Duke",2.5)

obj1.display()

print(obj1.No_tyre)
print(obj1.bike_name)
print(obj1.price)
```

### Access specifiers / types of attributes in encapsulation

```python
# 1.public : these variables can be accessed outside the class.
             # public variabes do not start with underscore.
# 2.protected : these variables can be accessed only inside/child class.
               # protected variable not INTENDED to access outside the class. 
               # proteced variable start with SINGLE underscore.
# 3.private : these variables can be accessed only inside the class.
               # not INTENDED to access outside/child class.
               # private variables starts with double underscore
```

# Using default arguments for function overloading

```python
# Ex-1:
def fun(var1,var2 = None,var3 = None):
    if var2 is not None:
        print(var1,var2,var3)
    else:
        print(var1)
fun(10)
fun(10,20,30)
```

```python
# Ex-2:default argument
def calculator(a = 0,b = 0,c = 0,d = 0):
    return a + b + c + d

print(calculator())
print(calculator(100))
print(calculator(100,200))
print(calculator(100,200,300))
print(calculator(100,200,300,400))
```

```python
# *args for function overloading

def calculator(*args):
    return sum(args)

print(calculator())
print(calculator(100))
print(calculator(100,200))
print(calculator(100,200,300))
print(calculator(100,200,300,400))
print(calculator(100,200,300,400,500,600,700,800))
```

```python
# function overloading using decorator
from multipledispatch import dispatch

@dispatch(int)
def fun(var1):
    print(var1)

@dispatch(int, int)
def fun(var1, var2):
    print(var1, var2)

@dispatch(float, float,int)
def fun(a, b, c):
    print(a, b, c)

fun(10)
fun(20, 30)
fun(1.2, 3.4, 5)
```

# 2. Operator overloading

```python
# we will perform different operations on the user defined objects using operators.
```

```python
class dosa:
    def __init__(self, variety,price):
        self.variety = variety
        self.price = price

    def __add__(self, d2_val):
        total_price = self.price + d2_val.price
        return total_price

    def __sub__(self, d2_val):
        total_price = self.price - d2_val.price
        return total_price

    def __mul__(self, d2_val):
        total_price = self.price * d2_val.price
        return total_price

    def __truediv__(self, d2_val):
        total_price = self.price / d2_val.price
        return total_price

    def __floordiv__(self, d2_val):
        total_price = self.price // d2_val.price
        return total_price

    def __mod__(self, d2_val):
        total_price = self.price % d2_val.price
        return total_price

    def __pow__(self, d2_val):
        total_price = self.price ** d2_val.price
        return total_price

d1 = dosa("Masala", 60)
d2 = dosa("Egg", 80)

print(d1 + d2) # __add__
print(d1 - d2) # __sub__
print(d1 * d2) # __mul__
print(d1 / d2) #__truediv__
print(d1 // d2) #__floordiv__
print(d1 % d2) #__mod__
# print(d1 ** d2) #__pow__
```

```python
# Comparision operator overloading

class dosa:
    def __init__(self, variety,price):
        self.variety = variety
        self.price = price

    def __gt__(self, d2_val):
        total_price = self.price > d2_val.price
        return total_price

    def __lt__(self, d2_val):
        total_price = self.price < d2_val.price
        return total_price

    def __eq__(self, d2_val):
        total_price = self.price == d2_val.price
        return total_price

    def __ge__(self, d2_val):
        total_price = self.price >= d2_val.price
        return total_price
        
    def __le__(self, d2_val):
        total_price = self.price <= d2_val.price
        return total_price

d1 = dosa("Masala", 60)
d2 = dosa("Egg", 80)

print(d1 > d2) # __gt__
print(d1 < d2) # __lt__
print(d1 == d2) # __eq__
print(d1 >= d2) #__ge__
print(d1 <= d2) #__le__


```

# Data Abstraction

```python
#Example-1: how to define abstract class, abstract method

from abc import ABC, abstractmethod

#abstract class
class Head_Mahindra(ABC):
    @abstractmethod
    def air_bag(self): #abstract method
        pass

    @abstractmethod
    def LED_Screen(self):
        pass

    def insurance(self): #concrete
        print("Please take Mahindra company insurance")
        

class hyd_mahindra(Head_Mahindra): # chid class
    def air_bag(self): #abstract method
        print("I will use Nylon to design the Air bag")

    def LED_Screen(self):
        print("Sony LED")

class noida_mahindra(Head_Mahindra):
    def air_bag(self): #abstract method
        print("I will use Cotton to design the Air bag")

    def LED_Screen(self):
        print("Samsung LED")


obj_hyd = hyd_mahindra() # create object of hyd child class
obj_noida = noida_mahindra()

obj_hyd.insurance()
# obj_head = Head_Mahindra()
```
