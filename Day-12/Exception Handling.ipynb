# Exception Handling


```python
# Def: EH allows ro respond to error without crashing the running application.
# with exceptuon handling,we can give our own error message.
```

```python
# Types of blocks in Exception handling:
* Try: Put the code where error can occurs.
* except: Handle the error.
* else: put code which you want to execute when their is No error.
* finally:execute either error or no error.
```

```python
#Example: 2

try:
    print(x)
except:
    print("some Error occur")

print("hello")
a,b = 10,20
print(a+b)
```

```python
# Example-3: structure of try-except

try:
    print(y)
except Exception as e:
    print(f"compiler error: {e}")

print("hello")
a,b = 10,20
print(a+b)
```

```python
# Example-2: structure of try-except

try:
    print(y)
except Exception as e:
    print(f"compiler error: {e}")
    print(type(e))

print("hello")
a,b = 10,20
print(a+b)
```

```python
# Example-2: else block: execute if NO error


try:
    print(y)
except Exception as e:
    print(f"compiler error: {e}")
    print(type(e))

else:
    print("Hurry, we successfully executed code")

print("hello")
a,b = 10,20
print(a+b)
```

```python
try:
    y = 20
    print(y)
except Exception as e:
    print(f"compiler error: {e}")
    print(type(e))

else:
    print("Hurry, we successfully executed code")

```

```python
# example-4 finally block: execute if NO error + Error
try:
    y = 20
    print(z)
except Exception as e:
    print(f"compiler error: {e}")
    print(type(e))

else:
    print("Hurry, we successfully executed code")
finally:
    print("I will execute whatever happen")

```

# type of errors


```python
# Name Error : variable not defined
#ZeroDivisionError
try:
    y = 10
    print(y/0)
except NameError:
    print("This is Name error")
except ZeroDivisionError:
    print("This is ZeroDivisionError")
```

```python
try:
    print(z)
except NameError:
    print("This is Name error")
except ZeroDivisionError:
    print("This is ZeroDivisionError")
```

```python
try:
    l1 = [10, 20, 30]
    print(l1[5])
except NameError:
    print("This is Name error")
except ZeroDivisionError:
    print("This is ZeroDivisionError")
except IndexError :
    print("This is IndexError")    
except Exception as e:
    print(type(e),e)
```

```python
try:
    d1 = {"name":"sreeja"}
    print(d1["age"])
except NameError:
    print("This is Name error")
except ZeroDivisionError:
    print("This is ZeroDivisionError")
except IndexError :
    print("This is IndexError") 
except KeyError :
    print("This is KeyError") 
except Exception as e:
    print(type(e),e)
```
