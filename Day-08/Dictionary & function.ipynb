# insert & update operations on dictionary


```python
x_dict1 = {"name" : "sandhya" , "age" : 23}

x_dict1["age"] = 24
print(x_dict1)

x_dict1['city'] = "bhpl"
print(x_dict1)

# insert and update using the update() function
# update() can insert and update multiple items together
x_dict1.update({"age" : 18,"occupation" : "IAS","name" : "sandhya", "salary" : 1.5})
print(x_dict1)
```

# Default parameters

```python
# assign a default value to the "var" variable
def fun(var = "Nothing"):
    print(var)

fun()
fun(100)
```

```python
def fun2(x = 10,y = 20):
    print(x + y)

fun2() # output 30
fun2(100) # output = 120
fun2(100,200)
```

# passing arbitrary number of arguments using *args

```python
def fun(*args):# in the place of args we use *var also
    for i in args:
        print(i)
fun()
fun("greeshuu")
fun("greeshu","sandhya")
```

# passing arbitrary key-value based argument using **Kargs

```python
def sreeja(**BF):
    for key in BF:
        print(key, " : ",BF[key])

sreeja(chaithanya = 150,varun = 250)
sreeja(chaithanya=150)
sreeja()
```

# lambda function

```python
# Def : a lambda function is a single line of code that is used to perform 
a short task without defining a full function with def.

# a lambda function don't have any name.
# althrough lambda function can receive arguments and process them.
# lambda function are popularaly used inside the other function 
to perform some small task
```

```python
obj = lambda a : a + 100

print(obj(10))
```

```python
obj = lambda a,b : a + b

print(obj(100,200))
```

```python
obj1 = lambda a,b,c : a*b*c

val = obj1(1,2,3)
print(val)
```

```python
# lambda function are popularaly used inside the other function 
 #to perform some small task

def fun(p):
    return lambda a : a**p

obj1 = fun(10)
print(obj1(2))
print(obj1(3))
print(obj1(4))
```

```python
# using lambda function inside the other function
def fun(p):
    obj = lambda a : a**p
    return obj

obj = fun(10)

print(obj1(2))
print(obj1(3))
print(obj1(4))
```

```python
def fun(p):
    obj = lambda a : a**p if a%2==0 else a+p
    return obj

obj = fun(2)

print(obj(2))
print(obj(3))
print(obj(4))
```

```python
# sort the below given list of tuple w,r,t marks

students = [("greeshma",90),("sangeetha",85),("rishi",83)]
students.sort()
print(students) # it is sorted based on the alphebits
```

```python
# sort the below given list of tuple w,r,t marks

students = [("greeshma",90),("sangeetha",85),("rishi",83)]

sorted_data = sorted(students,key = lambda a : a[1])
print(sorted_data) # it is sorted based on the alphebits and marks also
```

```python
students = [("greeshma",90,5),("sangeetha",85,2),("rishi",83,4)]

sorted_data = sorted(students,key = lambda a : a[2]) # it sorted second number
print(sorted_data)
```

```python
students = {"greeshma": 90,"sangeetha": 85,"rishi": 83}
data = students.items() # converting dictionary to list of tuple
print(data) # print(data)

sorted_data = sorted(data,key = lambda a : a[1])
print(sorted_data) 
```
