# for loop

```python
animals = ["pulli", "pilli", "kukka", "pandhi", "kothi"]
for i in animals:
    print(i)
```

```python
# for string type it give chachater by chachater
food = "chicken_bri"
for i in food:
    print(i)
```

```python
# for with the range()
# range() fun is used to generated list of the elements
#syntax of range: range(start, end, step)

for i in range(10):   #start=0, end = end -1, step =1
    print(i)
```

```python
for i in range(4,10):  #start=4, end = 10 -1, step =1
    print(i)
```

```python
# for loop automatically increment the i value

for i in range(4,10,2):      #start=4, end = 10 -1, step =2
    print(i)
```

```python
# example : break statement - to cmg out of the loop

for num in range(1,6):
    if num == 3:         # as n==3 so it will break at 3
        break
    print(num)
```

```python
#  example : continue statement - take the control to the start on the loop

for num in range(1,6):
    if num == 3:         # as n==3 so it will take control to start of loop at 3
        continue
    print(num)
```

```python
# example : nested loop
# for every value of outer loop, inner loop will completely execute

for i in range(1,4):
    for j in range(1,3):  # for 1 value of outer loop complete inner loop will execute
        print(i,j)
```

```python
# how many times the loop exeute its called as time complexitity for n = 3 , m= 3;time complexity = 15
```

```python
# example : else statement - executes if loop complete execution successfully

for i in range(5):
    if i == 2:
        continue
    elif i == 4:
        break
    else:
        print(i)
    pass
else:
    print("loop completed successfully")
```

```python
# example : pattern

for i in range(5):
    print("*", end= " ")
```

```python
for i in range(5):
    for j in range(5):
        print("*", end= " ")
    print()
```

```python
for i in range(1,6):
    for j in range(i):
        print("*", end = " ")
    print()
```

```python
for i in range(1,6):
    for j in range(6-i):
        print("*", end = " ")
    print()
```

```python
# other way
for i in range(5,0,-1):
    for j in range(i):
        print("*", end = " ")
    print()
```

```python
n= 5
for i in range(n,0,-1):
    print(" " * i + "* " * (n-i+1))
```

```python
n= 5
for i in range(n-1,0,-1):
    print(" " * i + "* " * (n-i+1))
```

# Operations on strings

### 1. Concatenation (+)

```python
str1 = "Babai "
str2 = "loves"
str3 = " Pinni"
str = str1 + str2 + str3
print(str)
```

### 2. Repetition (*)

```python
str = "bla "
print(str * 6)
```

### 3. Membership (in, not in)

```python
str = "HelloWorld"
print("Hell" in str)
print("Helll" in str)
print("Helo" not in str)
len(str)
```

### 4. Indexing & Slicing


```python
# we have negative and positive indexing
# string[start:end:step]  start = 0, end = n-1, step = 1

Heroine = "Samantha"

print(Heroine[4])
print(Heroine[-5])

print(Heroine[0:5])
print(Heroine[0:5:3])
print(Heroine[::])
print(Heroine[::2])
print(Heroine[6:3:1])   # if step size is positive we move to the right side & we r 

print(Heroine[6:3:-1])  # for -ve indexing we move left side and end = end - (-1) = end + 1
print(Heroine[-1 : -6])  # by default step size is 1 as +ve step we move right size & it print empty
print(Heroine[-1:-6:-1])
print(Heroine[-7::])     # by default step size is 1 as +ve step we move right size 
print(Heroine[-7::])

print(Heroine[0:-1:])    # combined +ve And -ve

print(Heroine[-10::])   # when we start at out of range we start from the last in ve
print(Heroine[-1:-6:-2])
print(Heroine[10:-6:])
print(Heroine[0:2:1] + Heroine[5::1])
```

```python
# write a code to reverse the string
print(Heroine[::-1])
print(Heroine[-1:-9:-1])
```

```python
# reverse the string
Heroine = "Samantha"

rev_result = ""

for char in Heroine:
    rev_result = char + rev_result
print(rev_result)
```
